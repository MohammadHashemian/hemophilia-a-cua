# Childhood OWSA scenario definitions | ages 1-15

This notebook inspects the horizon-aware base, low, and high one-way sensitivity scenarios. Scenario construction is reusable and lives in `app.notebook.owsa`.

```python
import polars as pl

from app.notebook.owsa.analysis import parameter_ranges
from app.notebook.owsa.workflow import load_owsa_inputs
from app.notebook.psa import CHILDHOOD
from app.notebook.scenario_helpers import parse_scenario

scenarios, base_parameters, ranges, context = load_owsa_inputs(CHILDHOOD)

assert all(parse_scenario(item.name)[0] == CHILDHOOD.key for item in scenarios)
assert all(item.overrides["cycles"].point() == CHILDHOOD.cycles for item in scenarios)
assert all(
    item.overrides["baseline_age"].point() == CHILDHOOD.start_age for item in scenarios
)

scenario_table = pl.DataFrame(
    {
        "scenario": [item.name for item in scenarios],
        "regime": [item.regime.value for item in scenarios],
        "extension": [parse_scenario(item.name)[3] for item in scenarios],
        "start_age": CHILDHOOD.start_age,
        "end_age": CHILDHOOD.end_age,
        "cycles": CHILDHOOD.cycles,
    }
)

with pl.Config(tbl_rows=-1, tbl_cols=-1, tbl_width_chars=1000, fmt_str_lengths=200):
    display(scenario_table)
    display(parameter_ranges(ranges))
```
