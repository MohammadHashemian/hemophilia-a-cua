# Childhood PSA simulation

Runs only the childhood horizon (ages 2–12) and writes to its dedicated cache.

```python
from app.notebook.psa import CHILDHOOD
from app.notebook.psa.workflow import run_horizon

results_path = run_horizon(CHILDHOOD)
results_path
```

```python
import polars as pl

df = pl.read_parquet(results_path)
assert df['time_horizon'].unique().to_list() == [CHILDHOOD.key]
assert df['cycles'].unique().to_list() == [CHILDHOOD.cycles]
assert df.group_by('scenario').agg(pl.col('iteration_id').n_unique()).select(pl.col('iteration_id').min()).item() == df.group_by('scenario').len()['len'].min()
df.group_by(['sampling_method', 'regime', 'extension']).len().sort(['sampling_method', 'regime', 'extension'])
```
