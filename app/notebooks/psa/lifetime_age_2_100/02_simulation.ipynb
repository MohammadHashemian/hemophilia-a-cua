# Lifetime PSA simulation

Runs only the lifetime horizon (ages 2–100) and writes to its dedicated cache.

```python
from app.notebook.psa import LIFETIME
from app.notebook.psa.workflow import run_horizon

results_path = run_horizon(LIFETIME)
results_path
```

```python
import polars as pl

df = pl.read_parquet(results_path)
assert df['time_horizon'].unique().to_list() == [LIFETIME.key]
assert df['cycles'].unique().to_list() == [LIFETIME.cycles]
assert df.group_by('scenario').agg(pl.col('iteration_id').n_unique()).select(pl.col('iteration_id').min()).item() == df.group_by('scenario').len()['len'].min()
df.group_by(['sampling_method', 'regime', 'extension']).len().sort(['sampling_method', 'regime', 'extension'])
```
