```python
import polars as pl
from app.notebook.psa import CHILDHOOD
from app.notebook.psa.workflow import load_horizon_results

df = load_horizon_results(CHILDHOOD)
assert df["time_horizon"].unique().to_list() == [CHILDHOOD.key]
assert df["cycles"].unique().to_list() == [CHILDHOOD.cycles]
assert (
    df.group_by("scenario")
    .agg(pl.col("iteration_id").n_unique())
    .select(pl.col("iteration_id").min())
    .item()
    == df.group_by("scenario").len()["len"].min()
)
grouped_scenarios = (
    df.group_by(["sampling_method", "regime", "extension"])
    .len()
    .sort(["sampling_method", "regime", "extension"])
)

with pl.Config(tbl_rows=-1, tbl_cols=-1, tbl_width_chars=1000, fmt_str_lengths=200):
    display(grouped_scenarios)
```
