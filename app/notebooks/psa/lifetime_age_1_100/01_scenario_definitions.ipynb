# Lifetime PSA scenario definitions

Scenarios begin at age 1 and run for 99 years, through age 100. This notebook inspects the authoritative definitions used by the simulation notebook.

```python
import polars as pl
from app.notebook.psa import LIFETIME
from app.notebook.psa.workflow import load_scenario_inputs
from app.notebook.scenario_helpers import parse_scenario

scenarios, base_params, context = load_scenario_inputs(LIFETIME)
assert all(parse_scenario(s.name)[0] == LIFETIME.key for s in scenarios)
assert all(s.overrides['cycles'].distribution.point() == LIFETIME.cycles for s in scenarios)

scenario_table = pl.DataFrame([
    {
        'scenario': s.name,
        'regime': s.regime.value,
        'start_age': LIFETIME.start_age,
        'end_age': LIFETIME.end_age,
        'cycles': LIFETIME.cycles,
        'sampling_method': parse_scenario(s.name)[2],
        'extension': parse_scenario(s.name)[3],
    }
    for s in scenarios
])
scenario_table
```
