# Childhood PSA scenario definitions

Scenarios begin at age 1 and run for 11 years, through age 12. This notebook inspects the authoritative definitions used by the simulation notebook.

```python
import polars as pl
from app.notebook.psa import CHILDHOOD_1_12
from app.notebook.psa.workflow import load_scenario_inputs
from app.notebook.scenario_helpers import parse_scenario

scenarios, base_params, context = load_scenario_inputs(CHILDHOOD_1_12)
assert all(parse_scenario(s.name)[0] == CHILDHOOD_1_12.key for s in scenarios)
assert all(
    s.overrides["cycles"].distribution.point() == CHILDHOOD_1_12.cycles
    for s in scenarios
)

scenario_table = pl.DataFrame(
    [
        {
            "scenario": s.name,
            "regime": s.regime.value,
            "start_age": CHILDHOOD_1_12.start_age,
            "end_age": CHILDHOOD_1_12.end_age,
            "cycles": CHILDHOOD_1_12.cycles,
            "sampling_method": parse_scenario(s.name)[2],
            "extension": parse_scenario(s.name)[3],
        }
        for s in scenarios
    ]
)
with pl.Config(tbl_rows=-1, tbl_cols=-1, tbl_width_chars=1000, fmt_str_lengths=200):
    display(scenario_table)
```
